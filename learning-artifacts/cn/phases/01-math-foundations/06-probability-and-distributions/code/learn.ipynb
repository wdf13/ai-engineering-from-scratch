## 1. 条件概率与独立性（已通过理解检查）

条件概率（$P(B)>0$）：

$$P(A\mid B)=\frac{P(A\cap B)}{P(B)}.$$

事件独立的判据：

$$P(A\cap B)=P(A)P(B).$$

掷公平骰子，$A=\{2,4,6\}$，$B=\{3,6\}$。计算：

$$P(A)=\tfrac12,\quad P(B)=\tfrac13,\quad P(A\cap B)=\tfrac16=P(A)P(B).$$

因此独立；学生在理解检查中正确选择 C。独立不等于互斥，两个概率均大于 0 的互斥事件不独立。


## 2. PMF、PDF 与 CDF

- **PMF**：离散随机变量的结果概率，$\sum_k p_X(k)=1$。
- **PDF**：连续分布的密度，区间概率来自积分：$P(u\le X\le v)=\int_u^v f_X(t)\,dt$。
- **CDF**：累计概率，$F_X(x)=P(X\le x)$。具有 PDF 的分布满足 $F_X(x)=\int_{-\infty}^x f_X(t)\,dt$。

对于 $X\sim U(a,b)$，且 $a<b$，在区间内部 PDF 为常数 $1/(b-a)$，CDF 为 $(x-a)/(b-a)$；区间外 PDF 为 0，CDF 左侧为 0，右侧为 1。PDF 在端点的取值不影响积分，本笔记按照教材采用闭区间实现。

**纠错记录**：学生最初提交的 `uniform_pdf` 区间内返回 `(x-a)/(b-a)`、在右端点之后返回 `1`，实际计算的是 CDF。因此保留该算法但更名为 `uniform_cdf`，并另写正确的 `uniform_pdf`。

学生对 $P(X=x_0)=0$ 的解释正确：单个点的积分区间宽度为零，$\int_{x_0}^{x_0}f_X(t)\,dt=0$。即使 $f_X(0.1)=5$，也不意味着点概率为 5。


## 3. Bernoulli PMF（学生原始实现，逻辑正确）

本题假定 $0\le p\le1$，暂不进行非法输入校验。

```python
import math

def bernoulli_pmf(k, p):
    if k == 0:
        return 1 - p
    elif k == 1:
        return p
    else:
        return 0
```

## 4. Uniform CDF（学生原算法更名，保留错误学习过程）

学生先前误将下列函数命名为 `uniform_pdf`，修正函数名称后可作为 CDF 使用。

```python
def uniform_cdf(x, a, b):
    if x <= a:
        return 0
    elif x > a and x <= b:
        return (x - a) / (b - a)
    else:
        return 1
```

## 5. Uniform PDF（导师给出的纠正实现）

下面是教学纠正代码，**不计作学生独立完成**。本题假定 $a<b$。

```python
def uniform_pdf(x, a, b):
    if a <= x <= b:
        return 1.0 / (b - a)
    return 0.0
```

## 7. Categorical PMF（学生已完成，经导师审核正确）

### 数学定义

一次试验的结果落入互斥且穷尽的类别集合 $\{0,1,\dots,K-1\}$ 中。

$$P(X=k)=p_k,\qquad p_k\ge 0,\quad \sum_{k=0}^{K-1} p_k=1.$$

学生解释：“分类的结果必定是一个类别，把所有的分类集合合起来概率就是 1。”思路正确；数学上还要求类别 **互斥**（每次试验仅发生一个类别），且这些类别 **穷尽** 全部可能结果。

### 学生原始实现

函数由学生独立提交。按本题前提，`probs` 是合法概率列表，`k` 是合法索引，所以直接索引是最简洁且正确的实现。

```python
def categorical_pmf(k, probs):
    # 返回类别 k 对应的概率
    # 假设 probs 是合法概率列表，且 k 是合法类别索引
    return probs[k]
```

```python
probs = [0.2, 0.5, 0.3]
assert math.isclose(categorical_pmf(0, probs), 0.2)
assert math.isclose(categorical_pmf(1, probs), 0.5)
assert math.isclose(categorical_pmf(2, probs), 0.3)
assert math.isclose(sum(categorical_pmf(k, probs) for k in range(len(probs))), 1.0)
print("Categorical PMF: 4 tests passed.")
```

## 8. Poisson PMF（学生已完成，代码审核正确）

Poisson 分布描述一定时间或空间内发生的**事件个数**（以适用的泊松过程假设为背景），参数 $\lambda>0$ 表示对应区间的理论平均事件数。

$$P(X=k)=\frac{\lambda^k e^{-\lambda}}{k!},\qquad k=0,1,2,\ldots$$

$$E[X]=\operatorname{Var}(X)=\lambda.$$

### 学生原始实现

保留学生代码原样。给定题目规定的合法输入，其计算公式正确。`math.exp(-lam)` 是可读性更好的等价选择，但不是必需修改。

```python
import math

def poisson_pmf(k, lam):
    # 计算泊松分布在 k 处的概率
    # 假设 lam > 0，k 为非负整数
    return (lam ** k * math.e ** -lam) / math.factorial(k)
```

### Poisson 测试与理解记录

在题目约定输入范围内，用 `math.exp` 的参考值进行交叉测试。以下运行输出来自导师执行环境，不冒充学生本地执行。

学生解释：若第一小时发生 4 次、第二小时发生 0 次，则样本均值为 2，但每小时次数可以波动。**评价：正确。**

进一步区分：$\lambda=2$ 是理论期望，`[4,0]` 的平均值 2 是样本均值。仅靠两个观测无法证明事件计数服从 Poisson(2)。若模型成立，$P(X=0)=e^{-2}\approx0.1353>0$。

```python
assert math.isclose(poisson_pmf(0, 2), math.exp(-2))
assert math.isclose(poisson_pmf(1, 2), 2 * math.exp(-2))
assert math.isclose(poisson_pmf(2, 2), 2 * math.exp(-2))
assert math.isclose(sum(poisson_pmf(k, 2) for k in range(30)), 1.0,
                    rel_tol=1e-12, abs_tol=1e-12)
print("Poisson PMF: 4 tests passed.")
print(f"P(X=0 | lambda=2) = {poisson_pmf(0, 2):.4f}")
```

## 9. 五种常见分布：理解它们描述的随机对象

| 分布 | 离散/连续 | 典型随机对象 | 参数 | 期望 | 方差 |
|---|---|---|---|---|---|
| Bernoulli | 离散 | 一次成败、一次二分类 | $p$ | $p$ | $p(1-p)$ |
| Categorical | 离散 | 一次试验落在 $K$ 个类别之一 | $p_0,\dots,p_{K-1}$ | 依类别数值编码而定 | 依类别数值编码而定 |
| Poisson | 离散 | 固定时间/空间内事件**次数** | $\lambda$ | $\lambda$ | $\lambda$ |
| Uniform | 连续 | 有限区间内各相同长度小区间等可能 | $a,b$ | $(a+b)/2$ | $(b-a)^2/12$ |
| Normal | 连续 | 连续波动、误差、基础高斯噪声 | $\mu,\sigma$ | $\mu$ | $\sigma^2$ |

注意：Bernoulli 是两类 Categorical 的特例；Poisson 的 $\lambda$ 既是平均次数也是方差；Uniform 与 Normal 用 PDF 而不是 PMF；连续随机变量在任意单点的概率为 0。Poisson 计数模型的适用性依赖事件发生机制的假设，不能单凭样本均值判断。

**概率模型与模型输出的联系**：二分类常使用 Bernoulli，多分类使用 Categorical；生成模型常以 Normal 噪声为起点。概率分布描述的是随机变量及其可能结果的规律，不意味着每次抽样都等于均值。

## 10. Normal PDF（学生独立实现，验收通过）

正态分布 $\mathcal N(\mu,\sigma^2)$ 的 PDF 是

$$f(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right),\qquad\sigma>0.$$

**学生的原始代码**（以下保留运算方式，新增换行以利阅读）符合上述公式。

**思考题记录**：学生认为增大标准差会让分布更分散、曲线更宽，且最高点下降；判断正确。需要修正学生文字中的括号：峰值是 $f(\mu)=1/(\sigma\sqrt{2\pi})$，**不是** \`1 / sigma * sqrt(2*pi)\`。后者的乘除优先级导致不同的值。$\sigma=1,2,4$ 时，峰值约为 $0.3989,0.1995,0.0997$。

理由：指数项在 $x=\mu$ 时达到最大值 1；密度积分始终为 1，因此分布变宽时峰值降低。

```python
def normal_pdf(x, mu, sigma):
    # mu: 均值
    # sigma: 标准差，假设 sigma > 0
    # 返回 x 位置的概率密度
    return 1 / (sigma * math.sqrt(2 * math.pi)) * math.exp(
        - (x - mu) ** 2 / (2 * (sigma ** 2))
    )
```

### Normal PDF 测试

导师用独立 Python 环境执行以下四项断言，结果 **4/4 通过**。本 Notebook 未冒充学生本地运行输出。

```python
assert math.isclose(normal_pdf(0, 0, 1), 1 / math.sqrt(2 * math.pi))
assert math.isclose(normal_pdf(1, 0, 1), math.exp(-0.5) / math.sqrt(2 * math.pi))
assert math.isclose(normal_pdf(0, 0, 2), 1 / (2 * math.sqrt(2 * math.pi)))
assert normal_pdf(2, 0, 1) < normal_pdf(1, 0, 1) < normal_pdf(0, 0, 1)
print("Normal PDF: 4 tests passed.")
```

## 11. 离散期望与方差（学生独立完成；导师验收通过）

**期望**：

$$E[X]=\sum_i x_i P(X=x_i).$$

**方差**：

$$\operatorname{Var}(X)=\sum_i P(X=x_i)(x_i-E[X])^2.$$

学生代码直接实现概率加权和以及围绕均值的平方偏差加权和。对于长度相同且合法的概率列表，公式与实现正确。

代码风格：可以用生成器表达式代替 \`sum([...])\` 以避免中间列表；原代码无逻辑错误。

**思考题记录**：学生答“方差为 0，则 X 是固定值”，正确。严谨表述为 $P(X=E[X])=1$（几乎必然为常数），因为非负随机变量 $(X-E[X])^2$ 的期望为 0，只能以概率 1 等于 0。

```python
def expected_value(values, probabilities):
    # 计算离散随机变量的理论期望
    return sum([v * p for v, p in zip(values, probabilities)])

def variance(values, probabilities):
    # 计算离散随机变量的理论方差
    e = expected_value(values,probabilities)
    return sum([p * (v - e) ** 2 for v, p in zip(values, probabilities)])
```

### 期望与方差测试（导师独立执行）

公平骰子、退化为常数的随机变量、非均匀离散分布共 6 项断言，**6/6 通过**。Notebook 未伪造学生本地执行结果。

```python
values = [1, 2, 3, 4, 5, 6]
probs = [1/6] * 6
assert math.isclose(expected_value(values, probs), 3.5)
assert math.isclose(variance(values, probs), 35/12)
assert math.isclose(expected_value([7], [1]), 7)
assert math.isclose(variance([7], [1]), 0)
assert math.isclose(expected_value([1, 9], [0.25, 0.75]), 7)
assert math.isclose(variance([1, 9], [0.25, 0.75]), 12)
print("Expected value and variance: 6 tests passed.")
```

## 12. Bernoulli 抽样（学生独立实现，导师审核通过）

对于独立伪随机数 $U_i\sim U[0,1)$，规则 $X_i=1$（当 $U_i<p$）、否则 $X_i=0$，便得到 Bernoulli($p$) 抽样。

下面保留学生提交的代码，不改变原算法。

```python
import random

def sample_bernoulli(p, n=1):
    # 根据 Bernoulli(p) 生成 n 个 0/1
    # 返回一个列表
    B = []
    for _ in range(n):
        if random.random() < p:
            B.append(1)
        else:
            B.append(0)
    return B
```

```python
random.seed(42)
assert sample_bernoulli(0, 100) == [0] * 100
assert sample_bernoulli(1, 100) == [1] * 100
samples = sample_bernoulli(0.7, 10000)
assert len(samples) == 10000
assert set(samples) <= {0, 1}
assert abs(sum(samples) / len(samples) - 0.7) < 0.03
print("Bernoulli sampling: 5 tests passed.")
print("Observed frequency:", sum(samples) / len(samples))
```

### 思考题纠正：单次方差 vs. 估计量方差

学生回答：“样本太小了，它的方差太大了，随着样本数的增多，方差会减小。”**方向正确**，需明确方差的对象。

对于独立同分布的 $X_i\sim\operatorname{Bernoulli}(p)$，单次方差 $\operatorname{Var}(X_i)=p(1-p)$ 与 $n$ 无关；样本频率 $\hat p=\sum_i X_i/n$ 的方差为

$$\operatorname{Var}(\hat p)=\frac{p(1-p)}{n}.$$

因此样本频率随 $n$ 增大通常更稳定。

## 14. 联合与边缘分布：理解检查（学生作答正确 3/3）

设 $X$ 为天气（0=晴，1=雨），$Y$ 为是否带伞（0=否，1=是），联合 PMF 为

|  | $Y=0$ | $Y=1$ | $P(X)$ |
|---|---:|---:|---:|
| $X=0$ | 0.40 | 0.10 | 0.50 |
| $X=1$ | 0.05 | 0.45 | 0.50 |
| $P(Y)$ | 0.45 | 0.55 | 1.00 |

学生回答：
1. 下雨的概率为 0.5 —— **正确**，$P(X=1)=0.05+0.45=0.50$。
2. 带伞的概率为 0.55 —— **正确**，$P(Y=1)=0.10+0.45=0.55$。
3. 天气与带伞不独立 —— **正确**，因为 $P(X=1,Y=1)=0.45$，但 $P(X=1)P(Y=1)=0.50\times0.55=0.275$。

补充：学生口头说“带伞加下雨”，数学上应说“**带伞且下雨**”，即交集，不能与事件概率的加法混淆。

条件概率复核：$P(Y=1\mid X=1)=0.45/0.50=0.90$，而 $P(Y=1)=0.55$，说明天气与带伞存在概率依赖关系。

### 数学定义

对于离散联合 PMF $p_{X,Y}(x,y)$：

$$p_X(x)=\sum_y p_{X,Y}(x,y),\qquad p_Y(y)=\sum_x p_{X,Y}(x,y).$$

行求和得到 $X$ 的边缘分布；列求和得到 $Y$ 的边缘分布。二者的概率和均应为 1。

## 15. 自己实现联合分布的边缘化（学生独立完成，导师审核通过）

输入二维合法联合 PMF（每行长度相等，总和为 1，非负），分别计算所有行和与列和。对于 $m\times n$ 联合表：

$$P(X=i)=\sum_{j=0}^{n-1}P(X=i,Y=j), \qquad P(Y=j)=\sum_{i=0}^{m-1}P(X=i,Y=j).$$

**学生原始提交代码**（没有改写核心循环和列表推导）：

```python
def joint_to_marginals(joint):
    # joint[i][j] 表示 P(X=i, Y=j)
    # 返回两个列表：marginal_x, marginal_y
    marginal_x = [sum(joint[i][j] for j in range(len(joint[i]))) for i in range(len(joint))]
    marginal_y = [sum(joint[i][j] for i in range(len(joint))) for j in range(len(joint[0]))]

    return marginal_x, marginal_y
```

### 测试与验收

导师在独立 Python 环境使用三个有效联合表检查了列表长度、边缘概率值、总和为 1，共 **15 项断言通过**。包含 2×2 非独立示例、2×3 非独立示例和 2×2 独立示例；本 Notebook 不伪造学生本地执行日志。

```python
import math

fixtures = [
    ([[0.40, 0.10], [0.05, 0.45]], [0.5, 0.5], [0.45, 0.55]),
    ([[0.10, 0.20, 0.20], [0.00, 0.30, 0.20]], [0.5, 0.5], [0.1, 0.5, 0.4]),
    ([[0.12, 0.18], [0.28, 0.42]], [0.3, 0.7], [0.4, 0.6]),
]
for joint, expected_x, expected_y in fixtures:
    mx, my = joint_to_marginals(joint)
    assert len(mx) == len(expected_x)
    assert len(my) == len(expected_y)
    assert all(math.isclose(a, b, abs_tol=1e-12) for a, b in zip(mx, expected_x))
    assert all(math.isclose(a, b, abs_tol=1e-12) for a, b in zip(my, expected_y))
    assert math.isclose(sum(mx), 1, abs_tol=1e-12) and math.isclose(sum(my), 1, abs_tol=1e-12)
print("Joint-to-marginals: 15 tests passed.")
```

### 思考题：归一化能否证明独立？

学生回答：“不能，这个只能说明它是一个正确的联合分布，跟独立没有关系。”判断**正确**。这里的前提是概率表的所有条目非负、且总和为 1。即使两个变量高度依赖，边缘概率仍分别归一化。独立需要对所有 $(i,j)$ 检查

$$P(X=i,Y=j)=P(X=i)P(Y=j).$$

## 16. 独立性检验（学生独立实现；已通过）

对任意合法的离散联合分布 $P(X=i,Y=j)$，随机变量 $X,Y$ 独立 **当且仅当所有** $(i,j)$ 都满足

$$P(X=i,Y=j)=P(X=i)P(Y=j).$$

为处理浮点数误差，可以使用绝对误差阈值 $\mathrm{tol}$。学生用已有的 \`joint_to_marginals()\` 计算边缘概率，对联合概率表全部单元格逐一检查。不满足时立即 \`return False\`，循环完全结束后才 \`return True\`。该控制流正确。

学生思考题回答：“只要有一处不独立他整体就不独立了，独立是相对于随机变量这一个整体。”**正确。** 数学上这是“全称命题”与“反例”的关系。

### 学生原始代码

```python
def check_independence(joint, tol=1e-9):
    # joint 是一个合法的二维联合概率表
    # 返回 True 或 False
    marginal_x, marginal_y = joint_to_marginals(joint)
    for i in range(len(joint)):
        for j in range(len(joint[i])): 
            if abs(joint[i][j] - marginal_x[i] * marginal_y[j]) > tol:
                return False

    return True
```

## 20. Categorical 抽样（学生独立实现，验收通过）

对于概率列表 $[p_0,\ldots,p_{K-1}]$，先令 $C_i=\sum_{j=0}^{i}p_j$；从 $U[0,1)$ 抽出一个随机数 $r$，找到第一个满足 $r<C_i$ 的类别索引 $i$。这是用累积概率划分均匀随机数区间的逆变换采样思路。

学生独立提交了嵌套的 \`sample()\` 函数，内部维护 \`cu\` 累计概率，按 \`r < cu\` 返回第一个命中的类别，外部列表推导重复 \`n\` 次。**算法正确**。

**思考题**：学生指出 \`int(random.random() * 3)\` 不是根据给定概率抽样；判断正确，因为它把 $[0,1)$ 分为三个等长区间，产生的类别分布是均匀的 $[1/3,1/3,1/3]$，不等于指定的 $[0.2,0.3,0.5]$。

**边界约定**：本题假设 \`probs\` 合法、非负、总和为 1，\`n\` 为正整数。浮点累计误差可能让最终 \`cu\` 极微小地低于 1，使极端随机数不落入任何区间，返回 \`None\`；本次测试没有遇到。工程版本可把最后一个类别作为兜底，并显式校验参数。

### 学生原始代码

```python
import random

def sample_categorical(probs, n=1):
    # 根据 probs 随机生成 n 个类别索引
    # 只能使用 random.random() 和基础 Python
    def sample():
        r = random.random()
        cu = 0
        for i in range(len(probs) ):
            cu += probs[i]
            if r < cu:
                return i
            
    return [sample() for _ in range(n)]
```

```python
random.seed(42)
assert sample_categorical([1, 0, 0], 50) == [0] * 50
assert sample_categorical([0, 0, 1], 50) == [2] * 50

samples = sample_categorical([0.2, 0.3, 0.5], 10000)
assert len(samples) == 10000
assert abs(samples.count(0) / 10000 - 0.2) < 0.03
assert abs(samples.count(1) / 10000 - 0.3) < 0.03
assert abs(samples.count(2) / 10000 - 0.5) < 0.03
print("Categorical sampling: 6/6 assertions passed.")
print("Observed frequencies:", [samples.count(k) / 10000 for k in range(3)])
```

## 21. Box–Muller 正态抽样：从数学原理推导（学生实现通过）

**背景**：已掌握 \`normal_pdf\`，但概率密度的计算不等于依据它抽样。如何只使用均匀随机数，产生独立的正态随机变量？

### 步骤 1：构造二维标准正态

令 $Z_1,Z_2$ 相互独立且都服从 $\mathcal N(0,1)$。因此联合密度是

$$f_{Z_1,Z_2}(z_1,z_2)=\frac{1}{2\pi}\exp\left[-\frac{z_1^2+z_2^2}{2}\right].$$

由于只有 $z_1^2+z_2^2$，联合密度具有旋转对称性。令

$$z_1=r\cos\theta,\quad z_2=r\sin\theta,\quad r=\sqrt{z_1^2+z_2^2}.$$

极坐标面积元素是 $r\,dr\,d\theta$（Jacobian 绝对值为 $r$）；因此

$$f_{R,\Theta}(r,\theta)=\frac{r}{2\pi}e^{-r^2/2}.$$

分解为 $[r e^{-r^2/2}]\,[1/(2\pi)]$，可知 $R$ 与 $\Theta$ 独立，角度在 $[0,2\pi)$ 均匀分布，而半径密度为 $f_R(r)=r e^{-r^2/2}$（$r\ge0$）。

**直觉**：半径为 $r$、厚度为 $dr$ 的薄圆环面积约为 $2\pi r\,dr$。半径概率密度比二维高斯多一个 $r$，来自圆环面积。

### 步骤 2：生成角度

如果 $U_2\sim U(0,1)$，则 $\Theta=2\pi U_2\sim U(0,2\pi)$。

### 步骤 3：用 CDF 反函数生成半径

$$F_R(r)=\int_0^r t e^{-t^2/2}\,dt=1-e^{-r^2/2}.$$

令 $U_1\sim U(0,1)$，求解 $F_R(R)=U_1$：

$$U_1=1-e^{-R^2/2}\Longrightarrow R=\sqrt{-2\ln(1-U_1)}.$$

因为 $U_1$ 与 $1-U_1$ 都均匀分布，公式一般记为 $R=\sqrt{-2\ln U_1}$。

### 步骤 4：恢复两个标准正态随机数

$$\boxed{Z_1=\sqrt{-2\ln U_1}\cos(2\pi U_2)}$$

$$\boxed{Z_2=\sqrt{-2\ln U_1}\sin(2\pi U_2)}$$

两个 $Z$ 均为独立的标准正态随机变量。只返回 $Z_1$ 完全有效，只是未利用可同时生成的 $Z_2$。

### 步骤 5：从标准正态映射到目标正态

对于 $Z\sim\mathcal N(0,1)$，定义 $X=\mu+\sigma Z$，则

$$E[X]=\mu+\sigma E[Z]=\mu,\qquad \operatorname{Var}(X)=\sigma^2\operatorname{Var}(Z)=\sigma^2.$$

**学生回答正确**：通过期望线性性与方差的平移不变、缩放平方性质证明。补全 $E[Z]=0$ 与 $\operatorname{Var}(Z)=1$ 即为完整推导。

### 学生独立提交的原始代码

核心公式与循环均正确。当前唯一明确发现的边界缺陷：当 `random.random()` 返回 0 时，`math.log(0)` 会抛出 `ValueError`。保留原代码作为学习证据。

```python
import math
import random

def sample_normal_box_muller(mu, sigma, n=1):
    # 使用 random.random() 产生均匀随机数
    # 使用 Box–Muller 变换生成正态随机数
    # 返回包含 n 个随机数的列表
    def Box_Muller():
        u1 = random.random()
        u2 = random.random()
        z = math.sqrt(-2 * math.log(u1)) * math.cos(2 * math.pi * u2)
        return z

    def sample(mu, sigma):
        return mu + sigma * Box_Muller()

    return [sample(mu, sigma) for _ in range(n)]
```

### 验证：导师在 Python 环境独立执行

`random.seed(42)`，从目标分布 $\mathcal N(2,9)$ 抽样 10000 次。平均值实测 **2.0027288621758244**，经验方差 **8.79136032648951**；5/5 项断言通过。导师另外强制 `random.random()` 返回 0，复现了 `ValueError('math domain error')`；这不是学生代码在普通采样时失败。

```python
random.seed(42)
normal_samples = sample_normal_box_muller(2, 3, 10000)
normal_mean = sum(normal_samples) / len(normal_samples)
normal_var = sum((x - normal_mean) ** 2 for x in normal_samples) / len(normal_samples)
assert len(normal_samples) == 10000
assert abs(normal_mean - 2) < 0.15
assert abs(normal_var - 9) < 0.6
assert len(sample_normal_box_muller(0, 1, 1)) == 1
assert all(math.isfinite(x) for x in normal_samples)
print("Normal sampling: 5/5 checks passed.")
print(f"Observed mean={normal_mean:.6f}, variance={normal_var:.6f}")
```

### 边界修正与工程实现

用 `u1=1.0-random.random()` 替代 `u1=random.random()`，可避免 `u1=0` 的数学定义域错误。下面是导师补充的参考实现，**不计为学生独立提交**。工程版本还应检查参数 `sigma>0` 和整数 `n>=1`。

```python
def sample_normal_box_muller_safe(mu, sigma, n=1):
    result = []
    for _ in range(n):
        u1 = 1.0 - random.random()
        u2 = random.random()
        z = math.sqrt(-2 * math.log(u1)) * math.cos(2 * math.pi * u2)
        result.append(mu + sigma * z)
    return result
```

## 23. Softmax：从 Logits 到合法的离散概率分布

### 为什么不能直接归一化 Logits？

三分类网络最后一层输出 $z=Wh+b$，例如 $z=[2,1,0]$ 代表猫、狗、鸟三个类别的原始分数（logits）。分数是任意实数，而概率 $p_i\ge0$ 且 $\sum_i p_i=1$。

直接使用 $z_i/\sum_jz_j$ 不可靠：例如 $z=[-2,0,2]$ 的总和为 0，$z=[-1,2,3]$ 的简单归一化还会得到负概率。

### Softmax 定义与性质

$$\boxed{p_i=\operatorname{softmax}(z)_i=\frac{e^{z_i}}{\sum_j e^{z_j}}}$$

- $e^{z_i}>0$ 保证正权重；
- 除以总和保证 $\sum_i p_i=1$；
- 相对概率满足 $p_i/p_j=e^{z_i-z_j}$，只依赖两个 logits 的差；
- 输出满足概率约束，但不保证模型对真实世界的预测已校准。

示例 $z=[2,1,0]$：

$$e^z=[7.3891,2.7183,1],\quad p\approx[0.665241,0.244728,0.090031].$$

Softmax 得到类别概率向量，随后可由此前手写的 \`categorical_pmf\` 查询单类概率，或用 \`sample_categorical\` 按概率抽样。

### 为什么使用指数：熵正则化推导

将 $p$ 约束在概率单纯形 $\Delta=\{p_i\ge0,\sum_i p_i=1\}$，求解：

$$\max_{p\in\Delta}\left[\sum_i p_i z_i-\sum_i p_i\log p_i\right].$$

第二项是概率分布的熵；若没有熵项，优化会将概率全部集中在分数最高的类别上。

引入拉格朗日乘子 $\lambda$：

$$\mathcal L=\sum_i p_i z_i-\sum_i p_i\log p_i+\lambda(\sum_i p_i-1).$$

驻点条件：

$$z_i-\log p_i-1+\lambda=0\Longrightarrow p_i\propto e^{z_i}.$$

结合 $\sum_i p_i=1$，恰好得到 Softmax。这是 Softmax 的一种优化视角，而非唯一的引入途径。

## 24. 为什么必须数值稳定：Max-Logit Trick

由于 \`math.exp(1002)\` 在通常的 Python 双精度环境下会发生 overflow，直接算 $z=[1000,1001,1002]$ 的 Softmax 不稳定。

Softmax 对整体平移不变：

$$\operatorname{softmax}(z_i-c)=\frac{e^{z_i-c}}{\sum_j e^{z_j-c}}=\operatorname{softmax}(z_i).$$

取 $m=\max(z)$，$z-m=[-2,-1,0]$，得到

$$\operatorname{softmax}([1000,1001,1002])\approx[0.090031,0.244728,0.665241].$$

现在所有指数输入都 $\le0$，避免正指数溢出；非常小的概率仍可能下溢为 0，所以涉及负对数概率时应继续使用对数域运算。

## 25. Log-Softmax、Log-Sum-Exp：在对数域计算概率

由

$$p_i=\frac{e^{z_i}}{\sum_j e^{z_j}}$$

取对数：

$$\log p_i=z_i-\log\sum_j e^{z_j}.$$

定义 $\mathrm{LSE}(z)=\log\sum_j e^{z_j}$。为避免指数溢出，令 $m=\max(z)$：

$$\boxed{\mathrm{LSE}(z)=m+\log\sum_j e^{z_j-m}}.$$

所以

$$\boxed{\operatorname{logsoftmax}(z)_i=z_i-\mathrm{LSE}(z)}.$$

**例子**：$z=[0,-1000]$，若先计算 Softmax，第二类概率可能下溢为 0，接着 \`log(0)\` 报错；直接使用 log-softmax 可稳定得到第二类对数概率约为 $-1000$。

对数概率还将连乘积转换为求和：$\log\prod_t p_t=\sum_t\log p_t$，是处理序列概率的常见方法。

## 26. Cross-Entropy / Negative Log Likelihood 与训练梯度

真实分类概率分布为 $q$，预测为 $p$，交叉熵为

$$H(q,p)=-\sum_iq_i\log p_i.$$

若 $q$ 是类别 $y$ 的 one-hot 标签，则

$$\boxed{L=-\log p_y=\mathrm{LSE}(z)-z_y}.$$

这提供了直接对 logits 进行稳定计算的形式。

对于 $z=[2,1,0]$，有 $\mathrm{LSE}(z)\approx2.407606$：

- 正确类别为 0：$L\approx0.407606$；
- 正确类别为 2：$L\approx2.407606$。

因此正确类别被分配的概率越小，损失越大。负对数损失在概率接近 0 时迅速增加。

**从 logits 导出梯度**：

$$\frac{\partial L}{\partial z_i}=\frac{e^{z_i}}{\sum_j e^{z_j}}-\mathbf1[i=y]=p_i-q_i.$$

当真类为 0 时：

$$\nabla_z L\approx[-0.334759,0.244728,0.090031].$$

梯度下降倾向于提高真类 logit 并降低其余类 logits；若要更新模型参数，还要通过 $z=Wh+b$ 继续链式求导。

**框架对照提示**：PyTorch 的 \`torch.nn.functional.cross_entropy(logits, target)\` 接受原始 logits，不需要先手动调用 Softmax。

## 27. 数值稳定 Softmax：学生独立实现，导师验收通过

学生正确地使用 \`m=max(logits)\` 以及 $e^{z_i-m}$ 来构造权重，归一化后返回概率列表。原始函数正确；可选优化是缓存计算过的指数列表，避免重复计算，但不作为错误处理。

**思考题**：学生回答“平移不变但通常不具有缩放不变性”，结论正确。严格理由：

$$e^{z_i+c}=e^c e^{z_i},\quad\frac{p_i}{p_j}=e^{z_i-z_j},\quad \frac{p_i'}{p_j'}=e^{a(z_i-z_j)}.$$

加公共常数时 $e^c$ 约掉；缩放 $a$ 改变 logit 差，故一般改变输出。这与 Temperature Softmax $\operatorname{softmax}(z/T)$ 相关：低温使分布更尖锐，高温使分布更平缓。

- $z=[2,1,0]$ 的概率约为 $[0.665241,0.244728,0.090031]$。
- $z=[102,101,100]$ 的概率相同。

```python
import math

def softmax(logits):
    # 输入：一个非空的实数列表
    # 输出：对应的类别概率列表
    # 要求：使用减去最大 logit 的稳定计算方法
    # 不调用现成的 softmax
    m = max(logits)
    s = sum(math.exp(logits[i] - m) for i in range(len(logits)))

    return [math.exp(logits[i] - m) / s for i in range(len(logits))]
```
