## 1. 条件概率与独立性（已通过理解检查）

条件概率（$P(B)>0$）：

$$P(A\mid B)=\frac{P(A\cap B)}{P(B)}.$$

事件独立的判据：

$$P(A\cap B)=P(A)P(B).$$

掷公平骰子，$A=\{2,4,6\}$，$B=\{3,6\}$。计算：

$$P(A)=\tfrac12,\quad P(B)=\tfrac13,\quad P(A\cap B)=\tfrac16=P(A)P(B).$$

因此独立；学生在理解检查中正确选择 C。独立不等于互斥，两个概率均大于 0 的互斥事件不独立。


## 2. PMF、PDF 与 CDF

- **PMF**：离散随机变量的结果概率，$\sum_k p_X(k)=1$。
- **PDF**：连续分布的密度，区间概率来自积分：$P(u\le X\le v)=\int_u^v f_X(t)\,dt$。
- **CDF**：累计概率，$F_X(x)=P(X\le x)$。具有 PDF 的分布满足 $F_X(x)=\int_{-\infty}^x f_X(t)\,dt$。

对于 $X\sim U(a,b)$，且 $a<b$，在区间内部 PDF 为常数 $1/(b-a)$，CDF 为 $(x-a)/(b-a)$；区间外 PDF 为 0，CDF 左侧为 0，右侧为 1。PDF 在端点的取值不影响积分，本笔记按照教材采用闭区间实现。

**纠错记录**：学生最初提交的 `uniform_pdf` 区间内返回 `(x-a)/(b-a)`、在右端点之后返回 `1`，实际计算的是 CDF。因此保留该算法但更名为 `uniform_cdf`，并另写正确的 `uniform_pdf`。

学生对 $P(X=x_0)=0$ 的解释正确：单个点的积分区间宽度为零，$\int_{x_0}^{x_0}f_X(t)\,dt=0$。即使 $f_X(0.1)=5$，也不意味着点概率为 5。


## 3. Bernoulli PMF（学生原始实现，逻辑正确）

本题假定 $0\le p\le1$，暂不进行非法输入校验。

```python
import math

def bernoulli_pmf(k, p):
    if k == 0:
        return 1 - p
    elif k == 1:
        return p
    else:
        return 0
```

## 4. Uniform CDF（学生原算法更名，保留错误学习过程）

学生先前误将下列函数命名为 `uniform_pdf`，修正函数名称后可作为 CDF 使用。

```python
def uniform_cdf(x, a, b):
    if x <= a:
        return 0
    elif x > a and x <= b:
        return (x - a) / (b - a)
    else:
        return 1
```

## 5. Uniform PDF（导师给出的纠正实现）

下面是教学纠正代码，**不计作学生独立完成**。本题假定 $a<b$。

```python
def uniform_pdf(x, a, b):
    if a <= x <= b:
        return 1.0 / (b - a)
    return 0.0
```

## 7. Categorical PMF（学生已完成，经导师审核正确）

### 数学定义

一次试验的结果落入互斥且穷尽的类别集合 $\{0,1,\dots,K-1\}$ 中。

$$P(X=k)=p_k,\qquad p_k\ge 0,\quad \sum_{k=0}^{K-1} p_k=1.$$

学生解释：“分类的结果必定是一个类别，把所有的分类集合合起来概率就是 1。”思路正确；数学上还要求类别 **互斥**（每次试验仅发生一个类别），且这些类别 **穷尽** 全部可能结果。

### 学生原始实现

函数由学生独立提交。按本题前提，`probs` 是合法概率列表，`k` 是合法索引，所以直接索引是最简洁且正确的实现。

```python
def categorical_pmf(k, probs):
    # 返回类别 k 对应的概率
    # 假设 probs 是合法概率列表，且 k 是合法类别索引
    return probs[k]
```

```python
probs = [0.2, 0.5, 0.3]
assert math.isclose(categorical_pmf(0, probs), 0.2)
assert math.isclose(categorical_pmf(1, probs), 0.5)
assert math.isclose(categorical_pmf(2, probs), 0.3)
assert math.isclose(sum(categorical_pmf(k, probs) for k in range(len(probs))), 1.0)
print("Categorical PMF: 4 tests passed.")
```

## 8. Poisson PMF（学生已完成，代码审核正确）

Poisson 分布描述一定时间或空间内发生的**事件个数**（以适用的泊松过程假设为背景），参数 $\lambda>0$ 表示对应区间的理论平均事件数。

$$P(X=k)=\frac{\lambda^k e^{-\lambda}}{k!},\qquad k=0,1,2,\ldots$$

$$E[X]=\operatorname{Var}(X)=\lambda.$$

### 学生原始实现

保留学生代码原样。给定题目规定的合法输入，其计算公式正确。`math.exp(-lam)` 是可读性更好的等价选择，但不是必需修改。

```python
import math

def poisson_pmf(k, lam):
    # 计算泊松分布在 k 处的概率
    # 假设 lam > 0，k 为非负整数
    return (lam ** k * math.e ** -lam) / math.factorial(k)
```

### Poisson 测试与理解记录

在题目约定输入范围内，用 `math.exp` 的参考值进行交叉测试。以下运行输出来自导师执行环境，不冒充学生本地执行。

学生解释：若第一小时发生 4 次、第二小时发生 0 次，则样本均值为 2，但每小时次数可以波动。**评价：正确。**

进一步区分：$\lambda=2$ 是理论期望，`[4,0]` 的平均值 2 是样本均值。仅靠两个观测无法证明事件计数服从 Poisson(2)。若模型成立，$P(X=0)=e^{-2}\approx0.1353>0$。

```python
assert math.isclose(poisson_pmf(0, 2), math.exp(-2))
assert math.isclose(poisson_pmf(1, 2), 2 * math.exp(-2))
assert math.isclose(poisson_pmf(2, 2), 2 * math.exp(-2))
assert math.isclose(sum(poisson_pmf(k, 2) for k in range(30)), 1.0,
                    rel_tol=1e-12, abs_tol=1e-12)
print("Poisson PMF: 4 tests passed.")
print(f"P(X=0 | lambda=2) = {poisson_pmf(0, 2):.4f}")
```

## 9. 五种常见分布：理解它们描述的随机对象

| 分布 | 离散/连续 | 典型随机对象 | 参数 | 期望 | 方差 |
|---|---|---|---|---|---|
| Bernoulli | 离散 | 一次成败、一次二分类 | $p$ | $p$ | $p(1-p)$ |
| Categorical | 离散 | 一次试验落在 $K$ 个类别之一 | $p_0,\dots,p_{K-1}$ | 依类别数值编码而定 | 依类别数值编码而定 |
| Poisson | 离散 | 固定时间/空间内事件**次数** | $\lambda$ | $\lambda$ | $\lambda$ |
| Uniform | 连续 | 有限区间内各相同长度小区间等可能 | $a,b$ | $(a+b)/2$ | $(b-a)^2/12$ |
| Normal | 连续 | 连续波动、误差、基础高斯噪声 | $\mu,\sigma$ | $\mu$ | $\sigma^2$ |

注意：Bernoulli 是两类 Categorical 的特例；Poisson 的 $\lambda$ 既是平均次数也是方差；Uniform 与 Normal 用 PDF 而不是 PMF；连续随机变量在任意单点的概率为 0。Poisson 计数模型的适用性依赖事件发生机制的假设，不能单凭样本均值判断。

**概率模型与模型输出的联系**：二分类常使用 Bernoulli，多分类使用 Categorical；生成模型常以 Normal 噪声为起点。概率分布描述的是随机变量及其可能结果的规律，不意味着每次抽样都等于均值。

## 10. Normal PDF（学生独立实现，验收通过）

正态分布 $\mathcal N(\mu,\sigma^2)$ 的 PDF 是

$$f(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right),\qquad\sigma>0.$$

**学生的原始代码**（以下保留运算方式，新增换行以利阅读）符合上述公式。

**思考题记录**：学生认为增大标准差会让分布更分散、曲线更宽，且最高点下降；判断正确。需要修正学生文字中的括号：峰值是 $f(\mu)=1/(\sigma\sqrt{2\pi})$，**不是** \`1 / sigma * sqrt(2*pi)\`。后者的乘除优先级导致不同的值。$\sigma=1,2,4$ 时，峰值约为 $0.3989,0.1995,0.0997$。

理由：指数项在 $x=\mu$ 时达到最大值 1；密度积分始终为 1，因此分布变宽时峰值降低。

```python
def normal_pdf(x, mu, sigma):
    # mu: 均值
    # sigma: 标准差，假设 sigma > 0
    # 返回 x 位置的概率密度
    return 1 / (sigma * math.sqrt(2 * math.pi)) * math.exp(
        - (x - mu) ** 2 / (2 * (sigma ** 2))
    )
```

### Normal PDF 测试

导师用独立 Python 环境执行以下四项断言，结果 **4/4 通过**。本 Notebook 未冒充学生本地运行输出。

## 11. 离散期望与方差（学生独立完成；导师验收通过）

**期望**：

$$E[X]=\sum_i x_i P(X=x_i).$$

**方差**：

$$\operatorname{Var}(X)=\sum_i P(X=x_i)(x_i-E[X])^2.$$

学生代码直接实现概率加权和以及围绕均值的平方偏差加权和。对于长度相同且合法的概率列表，公式与实现正确。

代码风格：可以用生成器表达式代替 \`sum([...])\` 以避免中间列表；原代码无逻辑错误。

**思考题记录**：学生答“方差为 0，则 X 是固定值”，正确。严谨表述为 $P(X=E[X])=1$（几乎必然为常数），因为非负随机变量 $(X-E[X])^2$ 的期望为 0，只能以概率 1 等于 0。

```python
def expected_value(values, probabilities):
    # 计算离散随机变量的理论期望
    return sum([v * p for v, p in zip(values, probabilities)])

def variance(values, probabilities):
    # 计算离散随机变量的理论方差
    e = expected_value(values,probabilities)
    return sum([p * (v - e) ** 2 for v, p in zip(values, probabilities)])
```

### 期望与方差测试（导师独立执行）

公平骰子、退化为常数的随机变量、非均匀离散分布共 6 项断言，**6/6 通过**。Notebook 未伪造学生本地执行结果。

```python
values = [1, 2, 3, 4, 5, 6]
probs = [1/6] * 6
assert math.isclose(expected_value(values, probs), 3.5)
assert math.isclose(variance(values, probs), 35/12)
assert math.isclose(expected_value([7], [1]), 7)
assert math.isclose(variance([7], [1]), 0)
assert math.isclose(expected_value([1, 9], [0.25, 0.75]), 7)
assert math.isclose(variance([1, 9], [0.25, 0.75]), 12)
print("Expected value and variance: 6 tests passed.")
```

## 12. Bernoulli 抽样（学生独立实现，导师审核通过）

对于独立伪随机数 $U_i\sim U[0,1)$，规则 $X_i=1$（当 $U_i<p$）、否则 $X_i=0$，便得到 Bernoulli($p$) 抽样。

下面保留学生提交的代码，不改变原算法。

```python
import random

def sample_bernoulli(p, n=1):
    # 根据 Bernoulli(p) 生成 n 个 0/1
    # 返回一个列表
    B = []
    for _ in range(n):
        if random.random() < p:
            B.append(1)
        else:
            B.append(0)
    return B
```

```python
random.seed(42)
assert sample_bernoulli(0, 100) == [0] * 100
assert sample_bernoulli(1, 100) == [1] * 100
samples = sample_bernoulli(0.7, 10000)
assert len(samples) == 10000
assert set(samples) <= {0, 1}
assert abs(sum(samples) / len(samples) - 0.7) < 0.03
print("Bernoulli sampling: 5 tests passed.")
print("Observed frequency:", sum(samples) / len(samples))
```

### 思考题纠正：单次方差 vs. 估计量方差

学生回答：“样本太小了，它的方差太大了，随着样本数的增多，方差会减小。”**方向正确**，需明确方差的对象。

对于独立同分布的 $X_i\sim\operatorname{Bernoulli}(p)$，单次方差 $\operatorname{Var}(X_i)=p(1-p)$ 与 $n$ 无关；样本频率 $\hat p=\sum_i X_i/n$ 的方差为

$$\operatorname{Var}(\hat p)=\frac{p(1-p)}{n}.$$

因此样本频率随 $n$ 增大通常更稳定。

## 14. 联合与边缘分布：理解检查（学生作答正确 3/3）

设 $X$ 为天气（0=晴，1=雨），$Y$ 为是否带伞（0=否，1=是），联合 PMF 为

|  | $Y=0$ | $Y=1$ | $P(X)$ |
|---|---:|---:|---:|
| $X=0$ | 0.40 | 0.10 | 0.50 |
| $X=1$ | 0.05 | 0.45 | 0.50 |
| $P(Y)$ | 0.45 | 0.55 | 1.00 |

学生回答：
1. 下雨的概率为 0.5 —— **正确**，$P(X=1)=0.05+0.45=0.50$。
2. 带伞的概率为 0.55 —— **正确**，$P(Y=1)=0.10+0.45=0.55$。
3. 天气与带伞不独立 —— **正确**，因为 $P(X=1,Y=1)=0.45$，但 $P(X=1)P(Y=1)=0.50\times0.55=0.275$。

补充：学生口头说“带伞加下雨”，数学上应说“**带伞且下雨**”，即交集，不能与事件概率的加法混淆。

条件概率复核：$P(Y=1\mid X=1)=0.45/0.50=0.90$，而 $P(Y=1)=0.55$，说明天气与带伞存在概率依赖关系。

### 数学定义

对于离散联合 PMF $p_{X,Y}(x,y)$：

$$p_X(x)=\sum_y p_{X,Y}(x,y),\qquad p_Y(y)=\sum_x p_{X,Y}(x,y).$$

行求和得到 $X$ 的边缘分布；列求和得到 $Y$ 的边缘分布。二者的概率和均应为 1。

## 15. 自己实现联合分布的边缘化（学生独立完成，导师审核通过）

输入二维合法联合 PMF（每行长度相等，总和为 1，非负），分别计算所有行和与列和。对于 $m\times n$ 联合表：

$$P(X=i)=\sum_{j=0}^{n-1}P(X=i,Y=j), \qquad P(Y=j)=\sum_{i=0}^{m-1}P(X=i,Y=j).$$

**学生原始提交代码**（没有改写核心循环和列表推导）：

```python
def joint_to_marginals(joint):
    # joint[i][j] 表示 P(X=i, Y=j)
    # 返回两个列表：marginal_x, marginal_y
    marginal_x = [sum(joint[i][j] for j in range(len(joint[i]))) for i in range(len(joint))]
    marginal_y = [sum(joint[i][j] for i in range(len(joint))) for j in range(len(joint[0]))]

    return marginal_x, marginal_y
```

### 测试与验收

导师在独立 Python 环境使用三个有效联合表检查了列表长度、边缘概率值、总和为 1，共 **15 项断言通过**。包含 2×2 非独立示例、2×3 非独立示例和 2×2 独立示例；本 Notebook 不伪造学生本地执行日志。

```python
import math

fixtures = [
    ([[0.40, 0.10], [0.05, 0.45]], [0.5, 0.5], [0.45, 0.55]),
    ([[0.10, 0.20, 0.20], [0.00, 0.30, 0.20]], [0.5, 0.5], [0.1, 0.5, 0.4]),
    ([[0.12, 0.18], [0.28, 0.42]], [0.3, 0.7], [0.4, 0.6]),
]
for joint, expected_x, expected_y in fixtures:
    mx, my = joint_to_marginals(joint)
    assert len(mx) == len(expected_x)
    assert len(my) == len(expected_y)
    assert all(math.isclose(a, b, abs_tol=1e-12) for a, b in zip(mx, expected_x))
    assert all(math.isclose(a, b, abs_tol=1e-12) for a, b in zip(my, expected_y))
    assert math.isclose(sum(mx), 1, abs_tol=1e-12) and math.isclose(sum(my), 1, abs_tol=1e-12)
print("Joint-to-marginals: 15 tests passed.")
```
