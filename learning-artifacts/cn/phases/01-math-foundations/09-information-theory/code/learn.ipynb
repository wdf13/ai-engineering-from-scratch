## 官方课程教学顺序（未验收前仅为课程计划）

1. 从前课复习两道概念题（不计成绩，不泄露当前回顾题答案）。

3. Concept：信息量、熵、交叉熵、KL、联合/条件熵与互信息、标签平滑、对数似然、bits/nats、perplexity，每项均需要数学推导与完整数值例子。
4. Build It：官方 5 组从零实现 Python 代码，要求学生独立编写关键部分并实际运行验证。
5. Use It：NumPy 和 PyTorch 交叉熵对照、数值稳定性、概率分布校验。
6. 4 道官方 Exercises：26 字母的均匀与真实频率熵；指定 logits 的 CE；KL 不对称；序列 PPL 实现。
7. Ship It：应用官方输出 skill-information-theory.md，结合真实案例产出可检查的个人成果。
8. 官方 3 道 stage=post 测验逐题作答并评分；只有全部完成后登记 LEARNING.md。

**注意边界**：本课涉及的随机变量先限定为离散分布；连续随机变量的微分熵、互信息估计属于额外内容。对随机噪声、编码长度、互信息特征选择的陈述必须明确适用条件。
