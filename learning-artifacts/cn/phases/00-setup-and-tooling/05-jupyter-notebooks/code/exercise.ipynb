## Exercise 1: %timeit list vs NumPy
Lesson 00/05. 100_000 random numbers.

```python
import numpy as np
```

```python
%timeit [x for x in range(100_000)]
```

```python
%timeit [np.random.random() for _ in range(100_000)]
```

```python
%timeit np.random.rand(100_000)
```

## Exercise 2: Restart & Run All
读表、画图。必须从上到下能跑通。

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "model": ["Linear", "Random Forest", "Neural Net"],
    "accuracy": [0.72, 0.89, 0.94],
    "train_time_sec": [0.1, 2.3, 45.6],
})
df
```

```python
%matplotlib inline
ax = df.plot(x="model", y="train_time_sec", kind="bar", legend=False, title="accuracy")
plt.tight_layout()
plt.show()
```
